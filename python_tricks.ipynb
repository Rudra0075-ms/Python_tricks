```python
n=5

for i in range(1,n+1):
    print("* "* i)   #comma(,) is not used because Value after * must be an iterable, not int. If comma(,) is used then, it'll show TypeError.
```

```python
n=5
for i in range(n):
    for j in range(n-i-1):
        print(" ",end=" ")
    for k in range(2*i+1):
        print("*",end=" ")      
    print()        
```
